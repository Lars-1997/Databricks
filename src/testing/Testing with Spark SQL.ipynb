```sql
USE CATALOG workspace;
USE SCHEMA default;
```

```sql
CREATE TABLE test (
  id INT,
  name STRING
)
```

```sql
INSERT Into test
values (1, 'John'), (2, 'Mary'), (3, 'Jane');
SELECT * FROM test
```

```sql
describe history test
```

```sql
CREATE TABLE IF NOT EXISTS test2 as SELECT * FROM test;
describe history test2
```

```sql
CREATE TABLE IF NOT EXISTS test3 DEEP CLONE test;
DESCRIBE HISTORY test3;
```

```sql
CREATE TABLE IF NOT EXISTS test4 SHALLOW CLONE test;
DESCRIBE HISTORY test4;
```
